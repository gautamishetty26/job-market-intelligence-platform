```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

print("Libraries imported successfully!")
```

```python
df = pd.read_csv('/content/job_data_merged_1 2.csv')

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
```

## 2. Visualization Selection and Design

Different visualizations are selected according to the nature of the
data. Bar charts are used for category comparisons, pie charts are
used for proportions, and other suitable charts are used to identify
patterns and relationships in the job market.

```python
print(df.columns.tolist())
```

```python
plt.figure(figsize=(10,6))

category_counts = df['Category'].value_counts().head(10)

sns.barplot(
    x=category_counts.values,
    y=category_counts.index
)

plt.title('Top Job Categories')
plt.xlabel('Number of Job Postings')
plt.ylabel('Job Category')
plt.tight_layout()
plt.show()
```

### 🔎 Interpretation

The bar chart shows the distribution of job postings across different
job categories. The categories with larger bars represent a greater
number of postings in the selected dataset.

```python
plt.figure(figsize=(8,5))

workplace_counts = df['Workplace'].value_counts()

sns.barplot(
    x=workplace_counts.index,
    y=workplace_counts.values
)

plt.title('Job Opportunities by Workplace Type')
plt.xlabel('Workplace Type')
plt.ylabel('Number of Job Postings')
plt.tight_layout()
plt.show()
```

### 🔎 Interpretation

The visualization compares remote, on-site and hybrid job opportunities.
It helps identify the workplace arrangement that appears most frequently
in the selected job-market dataset.

```python
plt.figure(figsize=(10,6))

location_counts = df['Location'].value_counts().head(10)

sns.barplot(
    x=location_counts.values,
    y=location_counts.index
)

plt.title('Top 10 Job Locations')
plt.xlabel('Number of Job Postings')
plt.ylabel('Location')
plt.tight_layout()
plt.show()
```

### 🔎 Interpretation

The chart identifies the locations with the highest number of job
postings in the dataset. This provides an overview of where the
analyzed job opportunities are concentrated.

```python
plt.figure(figsize=(12,6))

sns.countplot(
    data=df,
    x='Category',
    hue='Workplace'
)

plt.title('Job Categories by Workplace Type')
plt.xlabel('Job Category')
plt.ylabel('Number of Job Postings')
plt.xticks(rotation=45)
plt.legend(title='Workplace')
plt.tight_layout()
plt.show()
```

### 🔎 Interpretation

This visualization compares workplace arrangements across job categories.
It helps identify whether different job categories are more commonly
associated with remote, on-site or hybrid work.
